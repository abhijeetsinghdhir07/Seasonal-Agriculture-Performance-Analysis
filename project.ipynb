```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Read the uploaded dataset
df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")

# Remove rows with missing important data
df = df.dropna(subset=["Rainfall_mm", "Soil_Moisture_pct", "Yield_Tonnes_Ha"])

# Check the first few rows to confirm it worked
df.head()
```

```python
# Group data by season and calculate averages
seasonal_metrics = df.groupby('Season')[['Yield_Tonnes_Ha', 'Profit_INR', 'Water_Used_m3', 'Rainfall_mm']].mean().reset_index()

# Display the calculated results
print(seasonal_metrics)
```

```python
plt.figure(figsize=(8, 5))

# Added 'hue' and 'legend=False' to clear the warning
sns.barplot(data=seasonal_metrics, x='Season', y='Profit_INR', hue='Season', palette='viridis', legend=False)

plt.title('Average Agricultural Profit by Season')
plt.ylabel('Profit (INR)')
plt.xlabel('Season')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=seasonal_metrics, x='Season', y='Water_Used_m3', hue='Season', palette='Blues', legend=False)
plt.title('Average Water Usage by Season')
plt.ylabel('Water Used (m³)')
plt.show()
```
